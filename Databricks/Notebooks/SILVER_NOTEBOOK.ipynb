####CREATING_SILVER_CUSTOMERS_TABLE


```python
bronze_customer_df = spark.table(
    "workspace.default.bronze_customers"
)
```

```python
silver_customer_df = bronze_customer_df.dropDuplicates(
    ["customer_id"]
)
```

```python
display(silver_customer_df)
```

```python
silver_customer_df.write \
    .mode("overwrite") \
    .saveAsTable(
        "workspace.default.silver_customers"
    )
```

####CREATING_SILVER_PRODUCTS_TABLE

```python
bronze_product_df = spark.table(
    "workspace.default.bronze_products"
)
silver_product_df = bronze_product_df.dropDuplicates(
    ["product_id"]
)
```

```python
from pyspark.sql.functions import col

silver_product_df = silver_product_df.filter(
    col("price") > 0
)
```

```python
silver_product_df.write \
    .mode("overwrite") \
    .saveAsTable(
        "workspace.default.silver_products"
    )
```

####CREATING_SILVER_ORDERS_TABLE

```python
silver_order_df = spark.table(
    "workspace.default.bronze_orders"
)
silver_order_df = silver_order_df.filter(
    col("quantity") > 0
)
silver_order_df = silver_order_df.filter(
    col("amount") > 0
)
silver_order_df = bronze_order_df.dropDuplicates(
    ["order_id"]
)
```

```python
silver_order_df.write \
    .mode("overwrite") \
    .saveAsTable(
        "workspace.default.silver_orders"
    )
```
