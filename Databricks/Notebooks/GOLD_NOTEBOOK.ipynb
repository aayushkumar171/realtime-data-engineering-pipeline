###GOLD_CUSTOMER_DIMENSION_TABLE

```python
silver_customer_df = spark.table(
    "workspace.default.silver_customers"
)
```

```python
from pyspark.sql.functions import monotonically_increasing_id

dim_customer_df = (
    silver_customer_df
    .withColumn(
        "customer_key",
        monotonically_increasing_id()
    )
)
```

```python
dim_customer_df = dim_customer_df.select(
    "customer_key",
    "customer_id",
    "customer_name",
    "customer_city",
    "customer_state"
)
```

```python
dim_customer_df.write \
    .mode("overwrite") \
    .saveAsTable(
        "workspace.default.dim_customer"
    )
```

###GOLD_PRODUCT_DIMENSION_TABLE

```python
silver_product_df = spark.table(
    "workspace.default.silver_products"
)
dim_product_df = (
    silver_product_df
    .withColumn(
        "product_key",
        monotonically_increasing_id()
    )
)
```

```python
dim_product_df = dim_product_df.select(
    "product_key",
    "product_id",
    "product_name",
    "category",
    "price"
)
dim_product_df.write \
    .mode("overwrite") \
    .saveAsTable(
        "workspace.default.dim_product"
    )
```

###GOLD_FACT_TABLE_USING_ORDERS-TABLE

```python
silver_order_df = spark.table(
    "workspace.default.silver_orders"
)
fact_orders_df = silver_order_df.select(
    "order_id",
    "customer_id",
    "product_id",
    "quantity",
    "amount"
)
```

```python
fact_orders_df.write \
    .mode("overwrite") \
    .saveAsTable(
        "workspace.default.fact_orders"
    )
```
